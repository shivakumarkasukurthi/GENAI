```python
import numpy as np
```

```python
# create a 4*4 array of all false value
np.zeros((4,4),dtype=bool)
```

```python
# create a 4*4 array of all true value
np.ones((4,4),dtype=bool)
```

```python
# create a 4*4 array of diagonal 1 and non diag 0
np.eye(4)
```

```python
#create a 4*4 array of diagonal True and non diag False
np.eye(4,dtype=bool)
```

```python
#create a 4*4 array of diagonal False and non diag True
np.logical_not(np.eye(4,dtype=bool))
```
